### Consignas

1. Utilizar el método [`strip`](https://docs.python.org/3/library/stdtypes.html#str.strip) de `string`'s para quitar el salto de línea final (`"\n"`) de cada elemento de la lista.

```python
# "domicilios" es una lista de strings, por lo que genero una nueva lista donde a cada elemento le quito el string "\n"
domicilios_v1 = [s.strip("\n") for s in domicilios]
domicilios_v1
```

2. Separar el número que representa el ID del domicilio, del `string` que representa al domicilio en sí. El resultado debe ser una lista anidada, donde cada elemento de la lista maestra esté compuesto por estos dos componentes. Utilizar expresiones regulares y el método `group` para hacerlo.

```python
# Importo la librería re
import re

# Creo una copia de la lista domicilios_v1 (lista original sin el salto de linea "\n" final) para poder modificarla
domicilios_v2 = domicilios_v1.copy()

# Genero un for loop que itere sobre los elementos de la lista domicilios_v2 separando, por un lado, el código de la dirección y, por el otro, la dirección propiamente dicha
for i in range(len(domicilios_v2)):
    domicilios_v2[i] = re.split("\d;", domicilios_v2[i], maxsplit=1)

# Inspeciono el nuevo dataframe
domicilios_v2
```

3. Crear un `DataFrame` de `pandas`, llamado `domicilios_df`, con las columnas `ID` y `string_crudo`. El contenido debe ser la lista anidada creada arriba.

```python
# Importo la librería pandas
import pandas as pd

# Creo un dataframe
domicilios_df = pd.DataFrame(domicilios_v2)
domicilios_df.rename({0:"ID", 1:"string_crudo"}, axis=1, inplace=True)

# Inspecciono el nuevo dataframe
domicilios_df
```

5. Empecemos primero analizando un `string_crudo` particular: `"B° SAN LUIS XV(15);C: 03;;M: G;"`

Escribir una expresión regular que contenga grupos (sub-expresiones regulares escritas entre paréntesis) y permita extraer los campos `barrio`, `manzana`, `casa` y `lote` del `string_crudo` anterior. Usarla para extraer estos campos en un diccionario. El resultado debería ser `{"barrio": "SAN LUIS XV(15)", "manzana": "G", "casa": "03"}`.

```python
# Importo la librería re
import re

# Defino string particular
string = "B° SAN LUIS XV(15);C: 03;;M: G;"

# Defino expresión regular
re = re.search(r"^B°.(?P<barrio>.*?);M:.(?P<manzana>.*?);C:.(?P<casa>.*?);(L:)?(?P<lote>.*?);$", string)

# Creo diccionario
dict_string = {}

# Completo dicionario
keys = ["barrio", "manzana", "casa", "lote"]
for k in keys:
    try:
        dict_string[k] = re.group(k)
    except AttributeError:
        dict_string[k] = "null"

# Inspecciono el diccionario
print(dict_string)
```
